# Filtering (Coditional Statement)

## Classic Style

```python
df[df['sales'] > 500].head()
```

## Query Style

`pandas.read_sql(sql, con, index_col=None, coerce_float=True, params=None, parse_dates=None, columns=None, chunksize=None, dtype_backend=<no_default>, dtype=None)`

```python
# Query
df.query('sales > 500').head()
```

## Multiple Conditions

```python
# memfilter dengan dua atau lebih kondisi
# ingat: untuk menghindari kesalahan penulisan ekspresi, gunakan tanda kurung

# gunakan bitwise operator
# & untuk and
# | untuk or
# ~ untuk not

# memfilter transaksi dengan sales > 100 dari kategori Furniture

df[(df['sales'] > 500) & (df['quantity'] > 10)].head()
```

```python
df.query('sales > 500 & quantity > 10').head()
```

# ~ (Not)

```python
df[~(df['category'] == 'Furniture')].head()
```

# Sort Values

```python
# ascending = True
df.sort_values('sales').head()
```

```python
# ascending = False
# cara mengisi argumen:
  # 1. Positional Argument
  # 2. Keyword Argument
df.sort_values('sales', ascending=False).head()
```

```python
def pengurangan(x, y):
  return x - y
```

```python
# positional argument
pengurangan(2, 1)
```

```python
# positional argument
pengurangan(1, 2)
```

```python
# keyword argument
pengurangan(y=1, x=2)
```

```python
# keyword vs positional
pengurangan(2, y=1)
```

## Sorting more than 1 column

```python
df.sort_values(['sales', 'quantity'], ascending=[True, False]).head()
```
