# Dictionary

```python
pegawai = {
	'Nama'	: 'Dicky',
	'Usia'		: 30,
	'Role'		: 'Data Analyst',
	'Dept'		: 'Marketing'
}
pegawai
```

```python
pegawai['Nama']
```

```python
pegawai['Nama'] = 'Ricky'
pegawai
```

```python
pegawai['Gaji'] = 10000000
```

```python
pegawai
```

```python
del pegawai['Gaji']
```

```python
pegawai
```

```python
pegawai.keys()
```

```python
pegawai.values()
```

```python
pegawai.items()
```

```python
for key, value in pegawai.items():
  print(f"{key} : {value}")
```
